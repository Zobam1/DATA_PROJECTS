importing libraries

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
```

importing dataset

```python
tracker_set = pd.read_excel("Documents/Transportation & Logistics Tracking Dataset.xlsx")
```

view dataset

```python
tracker_set
```

choosing the columns we need for this project, having studied the dataset in excel

```python
tracker_sets = tracker_set[["Gps Provider", "Shipment Type", "Booking Date", "Ontime","Transportation Distance (KM)", "Vehicle Type", "Driver Name", "Supplier Name",
"Material Shipped"]]
```

counting the null values per variable

```python
for column in tracker_sets.columns.to_list():
    nullval = tracker_sets[column].isnull().value_counts()
    print(nullval)
```

dropping the null values totally

```python
tracker_sets.dropna(subset = ["Transportation Distance (KM)", "Vehicle Type", "Driver Name"], inplace = True)
```

checking our new dataset

```python
for column in tracker_sets.columns.to_list():
    nullval = tracker_sets[column].isnull().value_counts()
    print(nullval)
```

```python
tracker_sets.info()
```

extracting month, day and hour from booking date

```python
tracker_sets["Booking Date"]
```

```python
tracker_sets["Month"] = tracker_sets["Booking Date"].dt.month_name()
tracker_sets["Day"] = tracker_sets["Booking Date"].dt.day_name()
tracker_sets["Hour"] = tracker_sets["Booking Date"].dt.hour
```

```python
tracker_sets
```

Creating a column where we split the Hour variable into morning, day and evening, and the Day variable into weekday and weekend.

```python
hoursplit = np.linspace(tracker_sets["Hour"].min(), tracker_sets["Hour"].max(), num = 4, endpoint = True)
hours = ["morning", "day", "evening"]
tracker_sets["Hours"] = pd.cut(tracker_sets["Hour"], bins = hoursplit, labels = hours, include_lowest = True)
```

```python
Days = []
for day in tracker_sets["Day"]:
    if day in ("Monday", "Tuesday", "Wednesday", "Thursday", "Friday"):
        Days.append("weekday")
    else:
        Days.append("weekend")
tracker_sets["Days"] = Days
```

```python
tracker_sets.head()
```

Creating dummy variables for these columns: Days, Hours, Gps Provider and Vehicle Type

```python
weekday_end = pd.get_dummies(tracker_sets["Days"], dtype = int)
```

```python
days_days = pd.get_dummies(tracker_sets["Hours"], dtype = int)
```

```python
gps_dummy = pd.get_dummies(tracker_sets["Gps Provider"], dtype = int)
```

```python
vehicle_dummy = pd.get_dummies(tracker_sets["Vehicle Type"], dtype = int)
```

```python
tracker_sets = pd.concat([tracker_sets, days_days, weekday_end, gps_dummy, vehicle_dummy], axis=1)
```

```python
tracker_sets.reset_index()
```

normalize the transportation distance variable

```python
mean = tracker_sets["Transportation Distance (KM)"].mean()
```

```python
std = tracker_sets["Transportation Distance (KM)"].std()
```

```python
z_scores = (tracker_sets["Transportation Distance (KM)"] - mean)/std
```

```python
tracker_sets["Transportation_z_scores"] = z_scores
```

extracting the columns that we now need for the project

```python
new_tracker_sets = tracker_sets.drop(columns = ["Shipment Type", "Gps Provider", "Booking Date", "Vehicle Type", "Driver Name", "Supplier Name", "Material Shipped",
                                           "Month", "Day", "Hour", "Days", "Hours","Transportation Distance (KM)"], axis = 1).reset_index()
```

removing the "index" column

```python
new_tracker_sets = new_tracker_sets.iloc[:,1:]
```

```python
new_tracker_sets.head()
```

extracting the target variable

```python
actual_y = new_tracker_sets["Ontime"]
```

extracting the independent variables

```python
x_variables = new_tracker_sets.drop(columns = ["Ontime"], axis = 1)
```

spliting the datasets

```python
x_train, x_test, y_train, y_test = train_test_split(x_variables,actual_y, train_size = 0.75, random_state = 1, stratify=actual_y)
```

creating a model and modifying the maximum iteration to include all the rows

```python
lg_model = LogisticRegression(max_iter = 700)
```

fitting the datasets

```python
lg_model.fit(x_train, y_train)
```

predicting the y variable for the train dataset

```python
predicted_y = lg_model.predict(x_train)
```

```python
accuracy_score(y_train, predicted_y)
```

predicting the y variable for the test dataset

```python
pred_y_test = lg_model.predict(x_test)
```

```python
accuracy_score(y_test, pred_y_test)
```

```python
from sklearn.metrics import f1_score
```

```python
print(f1_score(y_train, predicted_y, pos_label = "No"))
print(f1_score(y_train, predicted_y, pos_label = "Yes"))
```

The logistic regression model is obviously not working because it keeps giving wrong results. F1 score is obviosuly poor for the yes labels meaning that there is a false negative error. So we try Random Forest algorithm

```python
random_model = RandomForestClassifier()
```

```python
random_model.fit(x_train, y_train)
```

```python
ytrain_predict = random_model.predict(x_train)
```

```python
ytest_predict = random_model.predict(x_test)
```

```python
accuracy_score(y_train, ytrain_predict)
```

```python
print(f1_score(y_train, ytrain_predict, pos_label = "No"))
print(f1_score(y_train, ytrain_predict, pos_label = "Yes"))
```
